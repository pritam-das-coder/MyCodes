```python
lst = [2,3,4,4,4,7,2]
print(dict.fromkeys(lst))
```

```python
seen = set()
unique = []
for x in lst:
    if x not in seen:
        seen.add(x)
        unique.append(x)
print(unique)
```

```python
counter = 0

def attempt_modify():
    counter += 1  # Attempts to modify global, but creates local variable instead

print("Global counter before:", counter)
attempt_modify()
print("Global counter after:", counter)  # Remains 0 due to local scope

# To actually modify global, use global keyword
def modify_global():
    global counter
    counter += 1

modify_global()
print("Global counter after global modify:", counter)  # Now 1
```
